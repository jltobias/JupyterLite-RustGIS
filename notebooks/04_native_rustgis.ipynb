**Run in local Jupyter, not JupyterLite.** Install `rustgis` in a compatible local Python environment. These calls follow the [upstream Python guide](https://github.com/opengeos/RustGIS/blob/main/docs/python.md). The sample data belongs to RustGIS; the teaching GeoJSON in this repository is separate.

## Runtime check

The native binding cannot be loaded by the Pyodide browser kernel. Keep native work on your computer and export files deliberately.

```python
import platform
print(platform.python_version(), platform.platform())
```

## Open the RustGIS sample project

```python
import rustgis
session = rustgis.Session(sample=True)
session
```

## Render and inspect

```python
session.to_png('rustgis_sample.png')
from IPython.display import Image, display
display(Image(filename='rustgis_sample.png'))
```

## Run a documented geoprocessing command

The sample project has a `Cities` layer. The buffer distance follows the current RustGIS tool/project settings; check units before using it on another dataset.

```python
session.run('gp.run', tool='buffer', params={'input':'Cities','distance':100})
session.to_png('rustgis_city_buffers.png')
display(Image(filename='rustgis_city_buffers.png'))
```

## Inspect the result layer

The upstream quick start names the output `Cities_Buffer`. GeoPandas is an optional native dependency for tabular inspection.

```python
gdf = session.layer('Cities_Buffer').to_geopandas()
print(gdf.shape)
gdf.head()
```
