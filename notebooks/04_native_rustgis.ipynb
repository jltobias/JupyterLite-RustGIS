# Native RustGIS Python session

**Run locally in Jupyter, not in JupyterLite.** Install the upstream native package with `python -m pip install rustgis`. See the [RustGIS Python guide](https://github.com/opengeos/RustGIS/blob/main/docs/python.md).


```python
import rustgis

session = rustgis.Session(sample=True)
session
```

```python
session.to_png('rustgis_sample.png')
from IPython.display import Image, display
display(Image(filename='rustgis_sample.png'))
```

## Buffer the sample cities

The call below follows the command form shown in the [RustGIS README](https://github.com/opengeos/RustGIS#quick-start). The distance value is interpreted by the current tool and project settings; consult the upstream tool help before applying it to your own data.


```python
session.run('gp.run', tool='buffer', params={'input': 'Cities', 'distance': 100})
session.to_png('rustgis_city_buffers.png')
display(Image(filename='rustgis_city_buffers.png'))
```

The upstream sample project is used here because the native API's file import signatures can change. Consult the [upstream documentation](https://github.com/opengeos/RustGIS/blob/main/docs/python.md) before adapting the session to your own data.
