## A map is a set of choices

Inspect the shared values before discussing perspective. Extrusion is a visual encoding of `height_m`, not evidence that these are actual buildings.

```python
import json
from pathlib import Path
geo = json.loads(Path('data/demo_features.geojson').read_text())
rows = [(f['properties']['name'], f['geometry']['type'], f['properties']['height_m']) for f in geo['features']]
for row in rows: print(f'{row[0]:20} {row[1]:8} {row[2]:>3} m')
```

## Chart before map

A bar chart makes relative heights easier to compare than a perspective scene. Compare the chart with the 3D map.

```python
from viz import bars, points
display(HTML(bars([r[0] for r in rows if r[1]=='Polygon'], [r[2] for r in rows if r[1]=='Polygon'], 'Compare mapped heights', ' m', '#e89a5b')))
```

## A simple coordinate view

```python
display(HTML(points(geo['features'], 'Plan view of the five features')))
```

## Try RustGIS's own 2D and globe modes

The [RustGIS Web URL syntax](https://github.com/opengeos/RustGIS/blob/main/docs/installation.md#web-app) supports `?data=` and `?projection=globe`. These open the actual RustGIS application, subject to browser CORS and WebGL support.

```python
from urllib.parse import quote
data_url = 'https://jltobias.github.io/JupyterLite-RustGIS/data/demo_features.geojson'
for name, projection in [('RustGIS 2D', ''), ('RustGIS globe', '&projection=globe')]:
    url = 'https://web.rustgis.app/?blank&data=' + quote(data_url, safe='') + projection
    display(HTML(f'<p><a href="{url}" target="_blank" rel="noopener">{name} ↗</a></p>'))
```
